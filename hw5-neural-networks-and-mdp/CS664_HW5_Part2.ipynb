```python
def display_grid(loc, q, v, pi):
    x, y = loc

    # Fill in the code to display the output in the required format shown i
    for i in range(x-1, x+2):
        for j in range(y-1, y+2):
            if (i, j) in v:
              print(f"({i},{j})  V={v[(i,j)]:.2f}  pi={pi[(i,j)]}  Q={q[(i,j)]}")
```
