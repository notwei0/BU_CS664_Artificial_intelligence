```python
class PancakeProblemRevised(Problem):
    """A PancakeProblem the goal is always `tuple(range(1, n+1))`, where the
    initial state is a permutation of `range(1, n+1)`. An act is the command `Flip i` 
    of the top `i` pancakes that will be flipped."""
    
    def __init__(self, initial): 
        self.initial, self.goal = tuple(initial), tuple(sorted(initial))
    
    def actions(self, state): 
        return [f'Flip {i}' for i in range(2, len(state) + 1)]

    def result(self, state, i): 
        pieces = i.split()
        flip_count = int(pieces[1])
        return state[:flip_count][::-1] + state[flip_count:]
    
    def h(self, node):
        "The gap heuristic."
        s = node.state
        return sum(abs(s[i] - s[i - 1]) > 1 for i in range(1, len(s)))
```

```python
class PancakeProblemWithFlipCost(PancakeProblemRevised):
    def action_cost(self, s, action, s1):
        pieces = action.split()
        flip_count = int(pieces[1])
        return flip_count
```
