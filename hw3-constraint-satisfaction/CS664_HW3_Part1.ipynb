```python
"""
Variable A gets the value a
Variable B gets the value b
Is this a valid combination?
"""

def check_contraints(A, a, B, b):
    if A != B and a == b:
        return False
        
    # Dave wants to live to the left of Bob
    if (A, B) == ("Dave", "Bob") and a > b:
        return False
    if (A, B) == ("Bob", "Dave") and b > a:
        return False

    # bob does not want to live next to charlie
    if (A, B) == ("Bob", "Charlie") and abs(a - b) == 1:
        return False
    if (A, B) == ("Charlie", "Bob") and abs(b - a) == 1:
        return False

    #alice wants a house to the immediate left of bob
    if (A, B) == ("Alice", "Bob") and a != b - 1:
        return False
    if (A, B) == ("Bob", "Alice") and b != a - 1:
        return False

    # No violation - return True
    return True
        
    
```

```python
# True case
check_contraints("Dave", 2, "Bob", 4)
```

```python
# False case
check_contraints("Bob", 2, "Charlie", 3)
```

```python
# True case
check_contraints("Alice", 2, "Bob", 3)
```

```python
def NeighborsCSP(neighbors):
    """Make a CSP for the problem."""
    
    if isinstance(neighbors, str):
        neighbors = parse_neighbors(neighbors)
        
    domains = {"Alice":    [1,2,3,4], 
               "Bob":      [1,2,3,4],
               "Charlie":  [1,2,3],
               "Dave":     [1,3,4]
              }
               
    return CSP(list(domains.keys()), domains, neighbors, check_contraints)
```
