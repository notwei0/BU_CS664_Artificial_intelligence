```python
from ortools.sat.python import cp_model
```

```python
model = cp_model.CpModel()

# Create variables.
b = model.NewIntVar(1, 9, "b")
a = model.NewIntVar(0, 9, "a")
s = model.NewIntVar(0, 9, "s")
e = model.NewIntVar(0, 9, "e")
l = model.NewIntVar(0, 9, "l")
m = model.NewIntVar(0, 9, "m")
g = model.NewIntVar(1, 1, "g")

# Force all letters to take on different values.
model.AddAllDifferent(b, a, s, e, l, m, g)
```

```python
"""
Create carry variables. c0 is true if the first column of addends carries a 1, 
c2 is true if the second column carries a 1, and so on.
"""

c0 = model.NewBoolVar("c0")
c1 = model.NewBoolVar("c1")
c2 = model.NewBoolVar("c2")
c3 = model.NewBoolVar("c3")



# Column 0:
model.Add(c0 == g)

# Column 1:
model.Add(c1 + b + b == a + 10 * c0)

# Column 2:
model.Add(c2 + a + a == m + 10 * c1)

#column 3:
model.Add(c3 + s + l == e + 10 * c2)

#column 4:
model.Add(e + l == s + 10 * c3)
```

```python
# Solve model.
solver = cp_model.CpSolver()
if solver.Solve(model) == cp_model.OPTIMAL:
    print("Optimal solution found!")
    
print("b:", solver.Value(b))
print("a:", solver.Value(a))
print("s:", solver.Value(s))
print("e:", solver.Value(e))
print("l:", solver.Value(l))
print("m:", solver.Value(m))
print("g:", solver.Value(g))
```
