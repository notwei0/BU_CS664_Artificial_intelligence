```python
constraint_map = {('one_across', 'one_down')  : (0,0),
                  ('one_across', 'two_down')  : (2,0),
                  ('three_across', 'two_down'): (2,2),
                  ('three_across', 'one_down'): (0,2),
                  ('four_across', 'two_down') : (0,4)}
constraint_map
```

```python
for (k1,k2), (v1,v2) in list(constraint_map.items()):
    constraint_map[(k2,k1)] = (v2,v1)
```

```python
constraint_map
```

```python
"""
Variable A gets the value a
Variable B gets the value b
Is this a valid combination?
"""
def check_contraints(A, a, B, b):
    if A != B and a == b:
        return False
    if (A,B) in constraint_map:
        v1, v2 = constraint_map[(A,B)]
        return meet_at(a, v1, b, v2)
    else:
        return False
        
    
```

```python
def CrosswordCSP(neighbors):
    """Make a CSP for the problem of the given crossword."""
    
    if isinstance(neighbors, str):
        neighbors = parse_neighbors(neighbors)
        
    domains = {'one_across':   ['ant', 'big', 'bus', 'car', 'has'],
               'one_down':     ['book', 'buys', 'hold', 'lane', 'year'],
               'two_down':     ['ginger', 'search', 'syntax', 'symbol'],
               'three_across': ['book', 'buys', 'hold', 'lane', 'year'],
               'four_across':  ['ant', 'big', 'bus', 'car', 'has']
              }
               
               
    return CSP(list(domains.keys()), domains, neighbors, check_contraints)
```
