```python
import nltk
from nltk.corpus import names

import numpy as np
import random
```

```python
nltk.download('names')
nltk.download('inaugural')
```

```python
male_names    = names.words('male.txt')
female_names  = names.words('female.txt')
print(f"Initial Male names: {len(male_names)}, Initial Female names: {len(female_names)}")

common_names  = np.intersect1d(male_names, female_names)
print(f"Common names: {len(common_names)}")

male_names    = np.setdiff1d(male_names,common_names)
female_names  = np.setdiff1d(female_names,common_names)
print(f"Pruned Male names: {len(male_names)}, Pruned Female names: {len(female_names)}")
```

```python
labeled_names = (
    [(name, 'male') for name in male_names] +
    [(name, 'female') for name in female_names]
)
```

```python
def classify_names(gender_features = lambda x: {'name': x}, train_pct = 0.8):
   
    # change the random seed with last 4 digits of your BU id
    
    random.seed(6449)
    
    featuresets = [(gender_features(name), gender) 
               for (name, gender) in labeled_names]
    print("Given features", len(featuresets))

    unique_features = []
    # Fill in the code for unique features from featuresets
    pairs = [(tuple(features.items()), label) for (features, label) in featuresets]
    unique_set = set(pairs)
    unique_features = [(dict(items), label) for (items, label) in sorted(unique_set)]
    
    print("Unique features", len(unique_features))

    train_set_size = int(len(unique_features) * train_pct)
    
    train_set = random.sample(unique_features, train_set_size)
    test_set = [item for item in unique_features if item not in train_set]

    print("Training set size", len(train_set))
    print("Test set size", len(test_set))

    # Fill in the code - Invoke the classifier on the training set
    
    classifier = nltk.NaiveBayesClassifier.train(train_set)

    print(f"Classifier Accuracy on test set: {nltk.classify.accuracy(classifier, test_set):.2f}")
    print(f"Classifier Accuracy on entire set: {nltk.classify.accuracy(classifier, unique_features):.2f}")

    # Fill in the code - Show the top 15 informative features
    classifier.show_most_informative_features(15)
   

    # Fill in the code to populate the actual labels and observed labels for the test set 

    actual_labels, observed_labels = [], []
    for (features, label) in test_set:
        actual_labels.append(label)
        observed_labels.append(classifier.classify(features))
    
    # Fill in the code to print the confusion matrix for test set
    print("Confusion matrix of test set:")
    print(nltk.ConfusionMatrix(actual_labels, observed_labels))
    
    # Fill in the code to populate the actual labels and observed labels for all the unique features data set
    
    actual_labels, observed_labels = [], []
    for (features, label) in unique_features:
        actual_labels.append(label)
        observed_labels.append(classifier.classify(features))
        
    # Fill in the code to print the confusion matrix for entire set

    print("Confusion matrix of entire set:")
    print(nltk.ConfusionMatrix(actual_labels,observed_labels))

    # Last statement
    return classifier
    
```

```python
# Fill in the code and replace with your first name

def name_features_v1(word):
     return {'last_two' : word[-2:]}
```

```python
name_features_v1('Wei')
```

```python
result1 = classify_names(gender_features = name_features_v1);
```
